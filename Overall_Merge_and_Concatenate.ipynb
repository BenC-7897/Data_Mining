```python
import os # Provides functions for interacting with the operating system
import shutil # Provides high-level file operations such as copying and moving files

# 1. Canonical label mapping: Maps dataset-specific folder names to unified class labels
LABEL_MAP = {
    "plastic": "plastic",
    "paper": "paper",
    "cardboard": "cardboard",
    "metal": "metal",
    "glass": "glass",

    "food": "organic",
    "biological": "organic",
    "vegetation": "organic",

    "clothes": "textile",
    "shoes": "textile",
    "textile": "textile",

    "battery": "battery",

    "trash": "trash",
    "miscellaneous": "trash"
}

# 2. Dataset roots: Each directory contains subfolders named by class
DATASET_ROOTS = [
    "realwaste-main/RealWaste",
    "original",
    "standardized_256",
    "standardized_384",
    "dataset-resized"
]

# 3. Output folder: All images will be copied into this single folder
OUTPUT = "Waste_Merged"
os.makedirs(OUTPUT, exist_ok=True)

# 4. Per-class counters: Used to generate unique filenames and avoid overwriting files
counters = {}

# 5. Merge loop
for root in DATASET_ROOTS:
    dataset_name = os.path.basename(root) # Extract dataset name for provenance in filenames

    for folder in os.listdir(root): # Loop through class folders inside the dataset
        src_label = folder.lower()

        if src_label not in LABEL_MAP: # Skip folders that are not mapped to a canonical class
            continue

        canonical = LABEL_MAP[src_label] # Get the unified class label
        counters.setdefault(canonical, 0) # Initialise counter for this class if not already present

        class_path = os.path.join(root, folder) # Full path to the class folder
        if not os.path.isdir(class_path): # Skip if the path is not a directory
            continue

        for fname in os.listdir(class_path): # Loop through all files in the class folder
            if fname.lower().endswith((".jpg", ".jpeg", ".png")): # Process only image files

                # Print the image currently being processed
                print(os.path.join(class_path, fname))

                new_name = ( # Construct a unique filename:
                    f"{canonical}_"
                    f"{dataset_name}_"
                    f"{counters[canonical]:06d}"
                    f"{os.path.splitext(fname)[1]}"
                )

                shutil.copy( # Copy the image into the merged output folder
                    os.path.join(class_path, fname),
                    os.path.join(OUTPUT, new_name)
                )

                counters[canonical] += 1 # Increment the counter for this class
```
